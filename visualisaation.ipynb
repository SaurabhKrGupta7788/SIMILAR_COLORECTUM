## Visualize masks over CT

```python
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt


img_path = r"D:\Similiar_Colorectum\14004841\images\images\crc_CRC068_AbdoPelvis1p5.nii.gz"
mask_path = r"D:\Similiar_Colorectum\14004841\lms\lms\crc_CRC068_AbdoPelvis1p5.nii.gz"
img = nib.load(img_path).get_fdata()
mask = nib.load(mask_path).get_fdata()

z = img.shape[2] // 2  # middle slice

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(img[:,:,z], cmap="gray")
plt.title("CT")

plt.subplot(1,3,2)
plt.imshow(mask[:,:,z], cmap="jet")
plt.title("Mask")

plt.subplot(1,3,3)
plt.imshow(img[:,:,z], cmap="gray")
plt.imshow(mask[:,:,z], cmap="jet", alpha=0.5)
plt.title("Overlay")

plt.show()
```
